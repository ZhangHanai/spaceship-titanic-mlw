```python
!pip install catboost -i https://pypi.tuna.tsinghua.edu.cn/simple

import pandas as pd
import numpy as np
from catboost import CatBoostClassifier, Pool
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
import warnings
import os
warnings.filterwarnings('ignore')

print("Current code working path：", os.getcwd())
train = pd.read_csv(r"E:\\train.csv")
test = pd.read_csv(r"E:\\test.csv")
submission = pd.read_csv(r"E:\\sample_submission.csv")
print(f"✅ Data loaded successfully! Training set{train.shape} Test set{test.shape}")

def safe_feature_engineering(df):
    data = df.copy()

    data[['Deck', 'Cabin_Num', 'Side']] = data['Cabin'].str.split('/', expand=True)
    data[['Group', 'Id']] = data['PassengerId'].str.split('_', expand=True)
    spend_cols = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
    data['Total_Spending'] = data[spend_cols].sum(axis=1)
    data['Has_Spending'] = (data['Total_Spending'] > 0).astype(int)
    data['Family_Size'] = data.groupby('Group')['Group'].transform('count')
    data['Is_Alone'] = (data['Family_Size'] == 1).astype(int)
    return data

train_fe = safe_feature_engineering(train)
test_fe = safe_feature_engineering(test)
all_data = pd.concat([train_fe, test_fe], ignore_index=True)

num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck', 
            'Cabin_Num', 'Total_Spending', 'Family_Size']
cat_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'VIP', 'Deck', 'Side', 'Group', 'Id']

for col in num_cols:
    all_data[col] = pd.to_numeric(all_data[col], errors='coerce')
    all_data[col].fillna(all_data[col].median(), inplace=True)

for col in cat_cols:
    all_data[col].fillna(all_data[col].mode()[0], inplace=True)

train_df = all_data[all_data['Transported'].notna()].copy()
test_df = all_data[all_data['Transported'].isna()].copy()

drop_cols = ['PassengerId', 'Cabin', 'Name', 'Transported']
X = train_df.drop(drop_cols, axis=1)
y = train_df['Transported'].astype(int)
X_test = test_df.drop(drop_cols, axis=1)

cat_features = [X.columns.get_loc(col) for col in cat_cols]
print(f"Total number of characteristics：{X.shape[1]}, Number of training set samples：{X.shape[0]}, Total number of labels：{y.shape[0]}")

FOLDS = 5
SEEDS = [42, 123]
test_pred_final = np.zeros(len(X_test))

params = {
    'iterations': 7000,
    'learning_rate': 0.02,
    'depth': 8,
    'l2_leaf_reg': 5.0,
    'random_strength': 1.2,
    'loss_function': 'Logloss',
    'eval_metric': 'Accuracy',
    'early_stopping_rounds': 100,
    'verbose': 200,
    'use_best_model': True
}

for seed in SEEDS:
    print(f"\n========== Training seed：{seed} ==========")
    skf = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=seed)
    val_scores = []
    test_pred_probs = np.zeros(len(X_test))

    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
        X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
        y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]
        
        train_pool = Pool(X_tr, y_tr, cat_features=cat_features)
        val_pool = Pool(X_val, y_val, cat_features=cat_features)
        test_pool = Pool(X_test, cat_features=cat_features)
        
        model = CatBoostClassifier(**params, random_state=seed)
        model.fit(train_pool, eval_set=val_pool)
        
        val_pred = model.predict(val_pool)
        acc = accuracy_score(y_val, val_pred)
        val_scores.append(acc)
        print(f"The{fold+1}  verification accuracy: {acc:.5f}")
        
        test_pred_probs += model.predict_proba(test_pool)[:, 1] / FOLDS

    print(f"Seed {seed} 5 fold average verification accuracy：{np.mean(val_scores):.5f}")
    test_pred_final += test_pred_probs / len(SEEDS)


val_all_probs = np.array([])
val_all_y = np.array([])

skf_base = StratifiedKFold(n_splits=FOLDS, shuffle=True, random_state=42)
for fold, (train_idx, val_idx) in enumerate(skf_base.split(X, y)):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]
    model = CatBoostClassifier(**params, random_state=42)
    train_pool = Pool(X_tr, y_tr, cat_features=cat_features)
    val_pool = Pool(X_val, y_val, cat_features=cat_features)
    model.fit(train_pool, eval_set=val_pool)

    val_all_probs = np.concatenate([val_all_probs, model.predict_proba(val_pool)[:,1]])
    val_all_y = np.concatenate([val_all_y, y_val])

best_thres = 0.5
best_acc = 0
for thres in np.arange(0.42, 0.58, 0.001):
    temp_pred = (val_all_probs > thres).astype(int)
    current_acc = accuracy_score(val_all_y, temp_pred)
    if current_acc > best_acc:
        best_acc = current_acc
        best_thres = thres
print(f"\n🎯 Global optimal classification threshold is found：{best_thres:.3f}")
print(f"Maximum accuracy of corresponding verification set：{best_acc:.5f}")
final_pred = (test_pred_final > best_thres).astype(bool)
submission = pd.DataFrame({
    'PassengerId': test['PassengerId'],
    'Transported': final_pred
})
submission.to_csv('catboost_bugfixed_top1_submission.csv', index=False)
```
